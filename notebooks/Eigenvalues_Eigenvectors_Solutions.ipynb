## Question 1
Find eigenvalues/eigenvectors of `A=[[4,1],[2,3]]`.

```python
import numpy as np

A = np.array([[4,1],
              [2,3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:", eigenvalues)
print("Eigenvectors (columns):\n", eigenvectors)
```

### Display each eigenpair

```python
for i, lam in enumerate(eigenvalues):
    v = eigenvectors[:, i]
    print(f"Eigenvalue {lam:.6f}")
    print("Eigenvector:", v)
    print()
```

## Question 2
Find eigenvalues/eigenvectors of `A=[[4,1,0],[1,4,1],[0,1,4]]`, then sort by decreasing eigenvalue.

```python
A = np.array([[4,1,0],
              [1,4,1],
              [0,1,4]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Original eigenvalues:", eigenvalues)
print("Original eigenvectors:\n", eigenvectors)
```

```python
order = np.argsort(eigenvalues)[::-1]
sorted_values = eigenvalues[order]
sorted_vectors = eigenvectors[:, order]

print("Sorted eigenvalues:", sorted_values)
print("Sorted eigenvectors (columns):\n", sorted_vectors)

for i, lam in enumerate(sorted_values):
    print(f"lambda = {lam:.6f}, v = {sorted_vectors[:, i]}")
```

## Question 3
Diagonalize `A=[[2,1,0],[0,3,1],[0,0,4]]` and verify \(AP=PD\).

```python
A = np.array([[2,1,0],
              [0,3,1],
              [0,0,4]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
# Use columns of eigenvectors to form P, and eigenvalues in the same order for D.
P = eigenvectors
D = np.diag(eigenvalues)

print("P =\n", P)
print("\nD =\n", D)
```

```python
AP = A @ P
PD = P @ D

print("A @ P =\n", AP)
print("\nP @ D =\n", PD)
print("\nDiagonalization verified?", np.allclose(AP, PD))
```
