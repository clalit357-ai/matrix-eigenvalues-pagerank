## Question 1
Convert the 2-feature dataset to NumPy, compute feature means, and center it.

```python
import numpy as np

X = np.array([[2,4],
              [3,5],
              [4,6],
              [5,8],
              [6,9]], dtype=float)

print("X =\n", X)
```

```python
mean = np.mean(X, axis=0)
X_centered = X - mean

print("Feature means:", mean)
print("\nCentered data =\n", X_centered)
```

## Question 2
Center the same dataset, compute its covariance matrix, find its eigenpairs, sort the eigenvalues, and identify the first principal component.

```python
X = np.array([[2,4],
              [3,5],
              [4,6],
              [5,8],
              [6,9]], dtype=float)

mean = np.mean(X, axis=0)
X_centered = X - mean
cov_matrix = np.cov(X_centered, rowvar=False)

print("Centered data =\n", X_centered)
print("\nCovariance matrix =\n", cov_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

print("Eigenvalues:", eigenvalues)
print("Eigenvectors (columns):\n", eigenvectors)
print("\nFirst principal component:", eigenvectors[:,0])
```

## Question 3
Perform PCA on the 3-feature dataset, sort components, calculate explained variance, and project onto the first two components.

```python
X = np.array([[2,4,1],
              [3,5,2],
              [4,6,3],
              [5,8,4],
              [6,9,5]], dtype=float)

mean = np.mean(X, axis=0)
X_centered = X - mean
cov_matrix = np.cov(X_centered, rowvar=False)

eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

print("Mean:", mean)
print("\nCentered data =\n", X_centered)
print("\nCovariance matrix =\n", cov_matrix)
print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors (columns):\n", eigenvectors)
```

```python
explained_variance_ratio = eigenvalues / np.sum(eigenvalues)
explained_percentage = explained_variance_ratio * 100

print("Explained variance (%):", explained_percentage)

W = eigenvectors[:, :2]
projected = X_centered @ W

print("\nFirst two principal components:\n", W)
print("\nProjected data onto first two PCs =\n", projected)
```
